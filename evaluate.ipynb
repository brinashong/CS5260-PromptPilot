# Evaluation of experimental results

```python
import os
import json
```

## Define dir names

```python
exp_name = "exp3"
results_dir = f"results/{exp_name}"
save_path = f"results/{exp_name}.json"
```

## For experiments 1 and 2

```python
# Initialize accumulators and counter
total_clip = 0.0
total_temporal = 0.0
total_dynamic = 0.0
file_count = 0

# Walk through all subdirectories and files
for dirpath, _, filenames in os.walk(results_dir):
    for filename in filenames:
        if filename.endswith(".json"):
            json_path = os.path.join(dirpath, filename)
            try:
                with open(json_path, "r") as f:
                    data = json.load(f)
                    scores = data.get("scores", {})
                    total_clip += scores.get("clip_tva_score", 0.0)
                    total_temporal += scores.get("temporal_consistency", 0.0)
                    total_dynamic += scores.get("dynamic_degree", 0.0)
                    file_count += 1
            except Exception as e:
                print(f"Error reading {json_path}: {e}")

# Calculate averages
if file_count > 0:
    avg_scores = {
        "average_clip_tva_score": total_clip / file_count,
        "average_temporal_consistency": total_temporal / file_count,
        "average_dynamic_degree": total_dynamic / file_count,
        "files_counted": file_count
    }

    # Save to JSON file
    with open(save_path, "w") as f:
        json.dump(avg_scores, f, indent=4)
    print(f"Averages saved to {save_path}")
else:
    print("No JSON files found.")
```

## For experiment 3

```python
# Accumulators for entry 0 and 1
totals = {
    0: {"clip_tva_score": 0.0, "temporal_consistency": 0.0, "dynamic_degree": 0.0, "count": 0},
    1: {"clip_tva_score": 0.0, "temporal_consistency": 0.0, "dynamic_degree": 0.0, "count": 0},
}

# Walk and read each JSON file
for dirpath, _, filenames in os.walk(results_dir):
    for filename in filenames:
        if filename.endswith(".json"):
            json_path = os.path.join(dirpath, filename)
            try:
                with open(json_path, "r") as f:
                    data = json.load(f)
                    for i, entry in enumerate(data):
                        if i > 1:
                            break  # Ignore beyond second entry
                        scores = entry.get("scores", {})
                        totals[i]["clip_tva_score"] += scores.get("clip_tva_score", 0.0)
                        totals[i]["temporal_consistency"] += scores.get("temporal_consistency", 0.0)
                        totals[i]["dynamic_degree"] += scores.get("dynamic_degree", 0.0)
                        totals[i]["count"] += 1
            except Exception as e:
                print(f"Failed to load {json_path}: {e}")

# Calculate averages
averages = {}
for i in [0, 1]:
    count = totals[i]["count"]
    if count > 0:
        averages[f"entry_{i}_average"] = {
            "clip_tva_score": totals[i]["clip_tva_score"] / count,
            "temporal_consistency": totals[i]["temporal_consistency"] / count,
            "dynamic_degree": totals[i]["dynamic_degree"] / count,
            "count": count
        }

# Save result
with open(save_path, "w") as f:
    json.dump(averages, f, indent=4)
print(f"Averages saved to {save_path}")
```
