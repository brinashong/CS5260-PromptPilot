# Prepare DIV2K data subset

```python
import os
import yaml
from PIL import Image
from pathlib import Path
```

## Define dir names

```python
dataset_dir = "dataset"
processed_dataset_dir = "preprocessed_dataset"
config_dir = "configs/images"
```

## Resize images to 480x320

```python
def resize_to_fixed_size(img, target_width=480, target_height=320):
    # Resize image to the target size directly
    return img.resize((target_width, target_height), Image.LANCZOS)

def resize_images(input_folder, output_folder):
    for subdir, _, files in os.walk(input_folder):
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff')):
                input_path = os.path.join(subdir, file)
                with Image.open(input_path) as img:
                    resized = resize_to_fixed_size(img, 480, 320)

                    # Maintain subfolder structure in output
                    rel_path = os.path.relpath(subdir, input_folder)
                    output_subdir = os.path.join(output_folder, rel_path)
                    os.makedirs(output_subdir, exist_ok=True)

                    output_path = os.path.join(output_subdir, file)
                    resized.save(output_path)
```

```python
resize_images(dataset_dir, processed_dataset_dir)
```

## Create config files for images

```python
def create_yaml_for_image(input_path, output_folder, subdir, image_filename):
    # Define the YAML structure with the subdirectory included in the paths
    yaml_data = {
        'input_path': f"dataset/{subdir}/{image_filename}.png",  # Including the subdirectory
        'save_path': f"./results/{subdir}/{image_filename}",  # Including the subdirectory
        'text_prompt': [],
        'negative_prompt': ""
    }

    # Create the corresponding YAML file
    yaml_filename = f"{image_filename}.yaml"
    yaml_output_path = os.path.join(output_folder, subdir, yaml_filename)

    # Ensure the subdirectory exists in the output folder
    os.makedirs(os.path.dirname(yaml_output_path), exist_ok=True)

    with open(yaml_output_path, 'w') as yaml_file:
        yaml.dump(yaml_data, yaml_file, default_flow_style=False)
    
    print(f"YAML for {image_filename} created at {yaml_output_path}")

def generate_yaml_files(input_folder, output_folder):
    # Walk through all subdirectories in the input folder
    for subdir, _, files in os.walk(input_folder):
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff')):
                # Get the image filename without extension
                image_filename = Path(file).stem
                
                input_path = os.path.join(subdir, file)
                # Create YAML files, passing the subdir name to include in paths
                create_yaml_for_image(input_path, output_folder, os.path.basename(subdir), image_filename)
```

```python
generate_yaml_files(processed_dataset_dir, config_dir)
```
